```python
%%writefile practice81.cpp
// Practice 8 task 1: Реализация обработки массива на CPU с использованием OpenMP
// 1. Создайте массив данных размером `N` (например, `N = 1 000 000`).
// 2. Реализуйте функцию для обработки массива на CPU с использованием OpenMP. Умножить каждый элемент массива на 2.
// 3. Замерьте время выполнения обработки на CPU.

#include <iostream>     // Для ввода и вывода данных
#include <vector>       // Для использования динамического массива vector
#include <omp.h>        // Библиотека OpenMP для параллельных вычислений

using namespace std;    // Чтобы не писать std::

int main() {            // Основная функция
    const int N = 1000000;               // Размер массива (1 000 000 элементов)
    vector<float> data(N);               // Создание массива вещественных чисел

    // Инициализация массива
    for (int i = 0; i < N; i++) {
        data[i] = i * 1.0f;              // Заполняем массив значениями
    }

    double start_time = omp_get_wtime(); // Засекаем начальное время

    #pragma omp parallel for              // Параллельный цикл OpenMP
    for (int i = 0; i < N; i++) {
        data[i] = data[i] * 2.0f;         // Умножаем каждый элемент на 2
    }

    double end_time = omp_get_wtime();    // Засекаем конечное время

    // Вывод времени выполнения
    cout << "Время выполнения CPU (OpenMP): "
         << end_time - start_time << " с" << endl;

    // Проверка корректности (вывод первых 5 элементов)
    cout << "Первые 5 элементов после обработки:" << endl;
    for (int i = 0; i < 5; i++) {
        cout << data[i] << " ";
    }
    cout << endl;

    return 0;                             // Завершение программы
}
```
